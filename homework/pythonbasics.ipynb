```python
#Create an array of numbers from 20-60
import numpy as np
arr=np.array(range(20, 61))
print(arr)

```

```python
#Select only even numbers from the array
for i in np.arange(20,61):
    if i%2==0:
        print(i)
```

```python
#Compute the mean and standard deviation of the array
mean = np.mean(arr)
std_dev = np.std(arr)
print("Mean:", mean)
print("Standard Deviation:", std_dev)
```

# problem 2
Write a normalization function that scales all values to the range [0, 1] 
use the MinMax normalization = `(x - min) / (max - min)`, 
where min and max should be the minimum and maximum values of the list. 


Test it on `np.array([10, 20, 30, 40, 50])` 
the output should be `[0.0, 0.25, 0.5, 0.75, 1.0]`.

```python
ex=np.array([10,20,30,40,50])
max=ex.max()
min=ex.min()

for i in ex:
    print((i-min)/(max-min))
```

# problem 3
For the given Titanic dataset below, compute:
1. How many passengers are in the dataset?
2. What is the average fare paid?
3. How many passengers were in each class (pclass)?
4. What is the survival rate for male vs female passengers?
5. Which 3 columns have the most missing values?

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv')
df.head(100)

```

```python
df.describe(include='all')
```

```python
#Number of passengers in the dataset
print(f"Number of passengers: {df['sex'].count()}")
```

```python
#Average fare paid
print(f"Average fare paid: {df['fare'].mean()}")
```

```python
#How many passengers were in each class (pclass)
print(df['pclass'].value_counts())
```

```python
#What is the survival rate for male vs female passengers
df.groupby('sex')['survived'].mean()
```

```python
#Which 3 columns have the most missing values
missing_values = df.isnull().sum()
missing_values_sorted = missing_values.sort_values(ascending=False)
top_3_missing_columns = missing_values_sorted.head(3)
print("Top 3 columns with the most missing values:")
print(top_3_missing_columns)
```

# problem 4
For the Titanic dataset, get:
- all female passengers 
- who survived 
- AND are older than 30 yrs. 

Print the number of rows in the result.

```python
#All female passengers who survived and were older than 30 years old
female_passengers = df['sex'] == 'female'
female_passengers_survived=female_passengers & (df['survived'] == 1)
female_passengers_survived_30=female_passengers_survived & (df['age'] > 30)
print("Number of female passengers who survived and were over 30:", female_passengers_survived_30.sum())
```
